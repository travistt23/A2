## 📊 Transaction Dataset (Sample Data)

This dataset simulates real banking transactions including everyday spending and refunds. It will be used to test cleaning, analysis, and financial insight generation functions.

```python
import pandas as pd

# Sample transaction data for testing
sample_transactions = {
    'Date': ['2024-08-01', '2024-08-02', '2024-08-03', '2024-08-04', '2024-08-05'],
    'Amount': ['$45.50', '$12.00', '$89.95', '$3.50', '-$25.00'],
    'Category': ['Groceries', 'Transport', 'Entertainment', 'Coffee', 'Refund'],
    'Description': ['Woolworths', 'Opal Card', 'Concert Tickets', 'Campus Cafe', 'Returned Item']
}

# Create DataFrame
df_sample = pd.DataFrame(sample_transactions)

print("📊 Sample transaction dataset created:")
print(df_sample)
```

**📊 Data Cleaning Function**

```python
def load_and_clean_transaction_data(df):
    """
    Cleans transaction data for analysis.

    Steps:
    - Removes $ symbols
    - Converts Amount to numeric
    - Handles missing values if any
    """

    df = df.copy()

    # Remove dollar signs and commas
    df["Amount"] = df["Amount"].replace("[$,]", "", regex=True)

    # Convert to numeric
    df["Amount"] = df["Amount"].astype(float)

    return df


# Test it
clean_data = load_and_clean_transaction_data(df_sample)
print(clean_data)
```

**📊 Spending Analysist**

```python
def analyze_spending_patterns(df):
    """
    Analyze spending patterns and return key financial insights.
    """

    # Ensure we don't modify original data
    df = df.copy()

    # Group by category and sum amounts
    category_totals = df.groupby("Category")["Amount"].sum()

    # Total spending (net)
    total_spending = df["Amount"].sum()

    # Find highest spending category
    top_category = category_totals.idxmax()
    top_value = category_totals.max()

    # Build results dictionary
    analysis = {
        "total_spending": round(total_spending, 2),
        "category_totals": category_totals.round(2),
        "top_category": top_category,
        "top_category_value": round(top_value, 2)
    }

    return analysis


# Run test
analysis = analyze_spending_patterns(clean_data)

print("💰 Total Spending:", analysis["total_spending"])
print("\n📊 Category Breakdown:")
print(analysis["category_totals"])
print("\n🔥 Highest Spending Category:")
print(analysis["top_category"], analysis["top_category_value"])
```

**💡 Recommendations Section**





```python
def generate_financial_recommendations(analysis_data):
    """
    Generate simple financial recommendations based on spending patterns.
    """

    top_category = analysis_data["top_category"]
    top_value = analysis_data["top_category_value"]
    total = analysis_data["total_spending"]

    # Basic logic for insights
    percentage = (top_value / total) * 100 if total != 0 else 0

    recommendations = []

    # Core insight
    recommendations.append(
        f"Your highest spending category is {top_category}, "
        f"which accounts for {round(percentage, 1)}% of your total spending."
    )

    # Behaviour-based advice
    if percentage > 40:
        recommendations.append(
            "This is a high proportion of your spending. Consider setting a budget limit for this category."
        )
    else:
        recommendations.append(
            "Your spending is fairly balanced across categories."
        )

    # General advice
    recommendations.append("Track your expenses regularly to improve financial awareness.")
    recommendations.append("Separate essential and non-essential spending to improve budgeting.")

    return "\n\n".join(recommendations)


# Test it
recommendations = generate_financial_recommendations(analysis)

print("💰 FINANCIAL RECOMMENDATIONS\n")
print(recommendations)
```

```python
def savings_goal_calculator(goal_amount, monthly_saving):
    """
    Calculates time needed to reach a savings goal.
    """

    if monthly_saving <= 0:
        return "Monthly saving must be greater than 0."

    months = goal_amount / monthly_saving
    years = months / 12

    return {
        "goal": f"${goal_amount}",
        "monthly_saving": f"${monthly_saving}",
        "months_needed": round(months, 1),
        "years_needed": round(years, 2)
    }


# Test
result = savings_goal_calculator(1000, 200)
print(result)
```

**🟢 1. Data**

```python
import pandas as pd

def load_and_clean_transaction_data(file_path):
    """
    Load and clean transaction data for the Smart Finance Assistant
    """

    try:
        # Load data
        df = pd.read_csv(file_path)

        # Validate required columns
        required_cols = ["Date", "Amount", "Category", "Description"]
        for col in required_cols:
            if col not in df.columns:
                raise ValueError(f"Missing required column: {col}")

        # Clean Amount column (remove $, commas, convert to float)
        df["Amount"] = df["Amount"].replace("[$,]", "", regex=True).astype(float)

        # Handle missing values
        df = df.dropna()

        # Ensure date format is consistent
        df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

        # Remove invalid rows after date conversion
        df = df.dropna(subset=["Date"])

        return df

    except Exception as e:
        print(f"❌ Error loading or cleaning data: {e}")
        return None


# Test your function
# clean_data = load_and_clean_transaction_data("your_file.csv")
# print(clean_data)
```

**🟢 2. Analysis**

```python
def analyze_spending_patterns(df):
    """
    Analyze spending patterns and generate business insights
    """

    try:
        if df is None or df.empty:
            return {"error": "No data provided for analysis"}

        # Total spending
        total_spending = df["Amount"].sum()

        # Spending by category
        category_spending = df.groupby("Category")["Amount"].sum()

        # Percentage breakdown
        category_percentage = (category_spending / total_spending) * 100

        # Highest spending category
        top_category = category_spending.idxmax()
        top_value = category_spending.max()

        # Build structured result
        analysis = {
            "total_spending": round(total_spending, 2),
            "category_spending": category_spending.to_dict(),
            "category_percentage": category_percentage.round(2).to_dict(),
            "top_category": top_category,
            "top_value": round(top_value, 2)
        }

        return analysis

    except Exception as e:
        return {"error": str(e)}


# Test your function
# analysis = analyze_spending_patterns(clean_data)
# print(analysis)
```

**🟢 3. Intelligence layer**

```python
def generate_financial_recommendations(analysis_data):
    """
    Generate actionable financial recommendations based on spending analysis
    """

    try:
        if "error" in analysis_data:
            return "Cannot generate recommendations due to invalid analysis data."

        total = analysis_data["total_spending"]
        top_category = analysis_data["top_category"]
        top_value = analysis_data["top_value"]
        category_percentage = analysis_data["category_percentage"]

        recommendations = []

        # General insight
        recommendations.append(f"Your total spending is ${total}.")

        # Top spending category insight
        recommendations.append(
            f"Your highest spending category is {top_category} (${top_value})."
        )

        # Spending warning logic
        if category_percentage.get(top_category, 0) > 50:
            recommendations.append(
                f"⚠️ {top_category} makes up over 50% of your spending. Consider reducing expenses in this area."
            )

        # Specific category advice
        if top_category.lower() == "entertainment":
            recommendations.append("Try setting a monthly entertainment budget to control discretionary spending.")

        elif top_category.lower() == "groceries":
            recommendations.append("Consider planning meals and using shopping lists to reduce grocery overspending.")

        elif top_category.lower() == "coffee":
            recommendations.append("Small daily purchases like coffee can add up — consider limiting frequency.")

        # General financial advice
        recommendations.append("Track your spending weekly to improve financial awareness.")
        recommendations.append("Separate needs vs wants to improve budgeting decisions.")

        return "\n".join(recommendations)

    except Exception as e:
        return f"Error generating recommendations: {e}"


# Test your function
# recommendations = generate_financial_recommendations(analysis)
# print(recommendations)
```

## Custom Financial Tools

```python
def create_savings_calculator_tool(current_savings, monthly_contribution, target_amount):
    """
    Savings goal calculator tool for Smart Finance Assistant
    """

    try:
        # Input validation
        if monthly_contribution <= 0:
            return "Monthly contribution must be greater than 0."

        if target_amount <= current_savings:
            return "You already reached your savings goal!"

        # Core calculation
        remaining_amount = target_amount - current_savings
        months_needed = remaining_amount / monthly_contribution

        # Format result
        result = {
            "current_savings": current_savings,
            "target_amount": target_amount,
            "monthly_contribution": monthly_contribution,
            "remaining_amount": remaining_amount,
            "months_needed": round(months_needed, 1)
        }

        return result

    except Exception as e:
        return f"Error: {e}"


# Test tool
# print(create_savings_calculator_tool(1000, 200, 3000))
```

## Gradio UI Integration

```python
import gradio as gr
import pandas as pd
from hands_on_ai.chat import get_response
from hands_on_ai import rag

# -----------------------------
# 💬 AI CHAT FUNCTION (FIXED)
# -----------------------------
def chat_interface(message):
    if message is None or message.strip() == "":
        return "Please enter a question."

    response = get_response(message)
    return response


# -----------------------------
# 📊 SIMPLE FINANCE ANALYSIS
# -----------------------------
def process_finance(file):
    df = pd.read_csv(file)

    df["Amount"] = df["Amount"].replace("[$,]", "", regex=True).astype(float)

    total = df["Amount"].sum()
    by_category = df.groupby("Category")["Amount"].sum()

    return f"""
💰 Total Spending: ${total:.2f}

📊 Spending by Category:
{by_category.to_string()}
"""


# -----------------------------
# 🔍 RAG FUNCTION (FIXED)
# -----------------------------
def rag_interface(question, file):
    if file is None:
        return "Please upload a CSV file."

    df = pd.read_csv(file)
    df["Amount"] = df["Amount"].replace("[$,]", "", regex=True).astype(float)

    context = df.to_string()

    response = rag.ask(f"""
User Question: {question}

Transaction Data:
{context}
""")

    return response


# -----------------------------
# 🛠 SAVINGS TOOL
# -----------------------------
def savings_tool_ui(current, monthly, target):
    if monthly <= 0:
        return "Monthly contribution must be greater than 0."

    months_needed = (target - current) / monthly

    return f"""
🎯 Savings Plan

Current Savings: ${current}
Monthly Contribution: ${monthly}
Target Goal: ${target}

⏳ Time Needed: {months_needed:.1f} months
"""


# -----------------------------
# 🎨 GRADIO UI
# -----------------------------
def create_finance_assistant_ui():

    with gr.Blocks(title="Smart Finance Assistant") as demo:

        gr.Markdown("# 🏦 Smart Finance Assistant")
        gr.Markdown("AI-powered personal finance tracker and advisor")

        # ---------------- TAB 1 ----------------
        with gr.Tab("📊 Upload & Analyse"):
            file_input = gr.File(label="Upload CSV Transactions")
            analyse_btn = gr.Button("Analyse Spending")
            output = gr.Textbox(label="Results", lines=8)

            analyse_btn.click(process_finance, inputs=file_input, outputs=output)

        # ---------------- TAB 2 ----------------
        with gr.Tab("💬 Chat Assistant"):
            chat_input = gr.Textbox(
                label="Ask a Finance Question",
                placeholder="e.g. I spend too much on coffee, what should I do?",
                lines=5
            )

            chat_btn = gr.Button("Send")
            chat_output = gr.Textbox(label="AI Response", lines=8)

            chat_btn.click(chat_interface, inputs=chat_input, outputs=chat_output)

        # ---------------- TAB 3 ----------------
        with gr.Tab("🔍 RAG Assistant"):
            rag_file = gr.File(label="Upload CSV")
            rag_question = gr.Textbox(
                label="Ask about your spending",
                placeholder="e.g. Where am I overspending?",
                lines=4
            )

            rag_btn = gr.Button("Ask")
            rag_output = gr.Textbox(label="Answer", lines=8)

            rag_btn.click(rag_interface, inputs=[rag_question, rag_file], outputs=rag_output)

        # ---------------- TAB 4 ----------------
        with gr.Tab("🛠️ Savings Tool"):
            current = gr.Number(label="Current Savings")
            monthly = gr.Number(label="Monthly Contribution")
            target = gr.Number(label="Savings Goal")

            tool_btn = gr.Button("Calculate")
            tool_output = gr.Textbox(label="Result", lines=6)

            tool_btn.click(savings_tool_ui, inputs=[current, monthly, target], outputs=tool_output)

    return demo


# -----------------------------
# 🚀 LAUNCH APP
# -----------------------------
demo = create_finance_assistant_ui()
demo.launch()
```
